## HW7 สร้างตารางใหม่ ที่ค่าใน sex เป็น missing ทั้งหมด
- สรุปว่าทำไม record นั้นๆถึงเป็น missing
- .groupby
- create pandas table
- simple visualization
- save table

```python
import pandas as pd
```

```python
import io

with open('/content/confirmed-cases.csv', 'rb') as f:
    raw_bytes = f.read()
    decoded_string = raw_bytes.decode('utf-8', errors='ignore')
```

```python
data_covid = pd.read_csv(io.StringIO(decoded_string), low_memory=False)
data_covid
```

```python
missing_sex_df = data_covid[data_covid['sex'].isnull()]
missing_sex_df
```

```python
missing_sex = missing_sex_df
```

```python
missing_sex_df.groupby('province_of_onset').describe()
```

```python
print("Nationality distribution for missing sex records:")
print(missing_sex_df['nationality'].value_counts(dropna=False).head(10))

print("\nRisk distribution for missing sex records:")
print(missing_sex_df['risk'].value_counts(dropna=False).head(10))
```

```python
records_df = pd.DataFrame(missing_sex_df.groupby('province_of_onset').size(), columns=['No.'])
records_df
```

```python
data_covid.groupby(data_covid['sex'].isnull()).count()
```

```python
numeric_df_for_plot = missing_sex[['No.', 'age', 'nationality']].dropna()
pd.plotting.parallel_coordinates(numeric_df_for_plot, 'nationality');
```

```python
pd.plotting.scatter_matrix(missing_sex);
```

```python
announce_date_count = missing_sex[missing_sex['province_of_onset']=='กรุงเทพมหานคร'].groupby('announce_date').count()
announce_date_count
```

```python
announce_date_count.to_csv('announce_date_count.csv')
```
